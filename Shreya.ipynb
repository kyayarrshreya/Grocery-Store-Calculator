GROCERY STORE CALCULATOR - CASE STUDY

```python
print("Welcome to the Grocery Store Calculator!")

# Initialize an empty list to store items
items = []

while True:
    # Get item name
    item_name = input("Enter the name of the item (or type 'done' to finish): ")
    if item_name.lower() == 'done':
        break

    # Get item price
    while True:
        try:
            item_price = float(input(f"Enter the price of {item_name}: "))
            if item_price < 0:
                print("Price cannot be negative. Please enter a valid price.")
            else:
                break
        except ValueError:
            print("Invalid input. Please enter a numeric value for the price.")

    # Get item quantity
    while True:
        try:
            item_quantity = int(input(f"Enter the quantity of {item_name}: "))
            if item_quantity < 0:
                print("Quantity cannot be negative. Please enter a valid quantity.")
            else:
                break
        except ValueError:
            print("Invalid input. Please enter a numeric value for the quantity.")

    # Add item to the list
    items.append((item_name, item_price, item_quantity))

# Calculate total cost
total_cost = sum(price * quantity for _, price, quantity in items)

# Print the receipt
print("\n--- Receipt ---")
for item_name, item_price, item_quantity in items:
    print(f"{item_name}: {item_price:.2f} x {item_quantity} = {item_price * item_quantity:.2f}")

print(f"\nTotal Cost: {total_cost:.2f}")
print("Thank you for shopping with us!")
```
